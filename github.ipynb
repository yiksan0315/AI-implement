```python
import os
import sys
from google.colab import drive

drive.mount("/content/drive")

# 2. 레포지토리 정보 설정
USERNAME = "yiksan0315"
REPO_NAME = "AI-implement"
PROJECT_PATH = f"/content/drive/My Drive/{REPO_NAME}"

# 인증 정보가 포함된 URL 주소 생성
REPO_URL = f"https://github.com/{USERNAME}/{REPO_NAME}.git"

# 3. 최초 1회는 Clone, 이미 존재하면 Pull 수행
if not os.path.exists(PROJECT_PATH):
    %cd "/content/drive/My Drive/"
    !git clone {REPO_URL}
    print(f"[{REPO_NAME}] cloned.")
else:
    %cd {PROJECT_PATH}
    !git pull
    print(f"[{REPO_NAME}] pulled.")

# 4. 파이썬이 이 폴더 안의 모듈(.py)을 찾을 수 있도록 경로 주입
if PROJECT_PATH not in sys.path:
    sys.path.insert(0, PROJECT_PATH)
```
